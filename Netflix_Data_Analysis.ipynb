# DATA CLEANING

# 9. Check Missing Values

```python
# Check the number of missing values in each column
df.isnull().sum()
```

# 14. Convert Release Date

```python
# Convert Release_Date column into datetime format

df['Release_Date'] = pd.to_datetime(df['Release_Date'], errors='coerce')

# Display the result
df[['Title', 'Release_Date']].head()
```

# 15. Create Release Year

```python
# Extract release year from Release_Date
df['Release_Year'] = df['Release_Date'].dt.year

# Display the result
df[['Title', 'Release_Date', 'Release_Year']].head()
```

# GENRE ANALYSIS

# 35. Check Genre Data

```python
# Display sample genre values

df['Genre'].head(10)
```

# 36. Count Genre

```python
# Split multiple genre and count individual genre

genres = (df['Genre'].dropna().str.split(',').explode())

# Remove extra spaces
genres = genres.str.strip()

#  Count genres
genre_counts = genres.value_counts()

genre_counts.head(10)
```

# 37.  Top 10 Genre Visualization

```python
top_genres = genre_counts.head(10)

plt.figure(figsize=(12,6))

sns.barplot(x=top_genres.values, y=top_genres.index, color='steelblue')

plt.title("Top 10 Movie Genres", fontsize=16)

plt.xlabel("Number of Movies")
plt.ylabel("Genre")

plt.show()
```

# 38. Genre-wise Average Rating

```python
# Create a dataframe for genre rating analysis

genre_df = df[['Genre', 'Vote_Average']].dropna()

# Split multiple genres

genre_df = genre_df.assign( Genre=genre_df['Genre'].str.split(',')).explode('Genre')

# Remove extras spaces

genre_df['Genre'] = genre_df['Genre'].str.strip()

# Calculate average rating by genre

genre_rating = ( genre_df.groupby('Genre')['Vote_Average'].mean().sort_values(ascending=False))

genre_rating.head(10)
```

# 39. Genre Rating Visualization

```python
# Visualization: Top 10 genres by average rating

top_genre_ratings = genre_rating.head(10)

plt.figure(figsize=(12,6))

sns.barplot( x=top_genre_ratings.values, y=top_genre_ratings.index, color='darkcyan')
plt.title("Top 10 Genres by Average Rating", fontsize=16)

plt.xlabel("Average Rating")
plt.ylabel("Genre")

plt.show()
```
